```python

import pandas as pd

# Create a sample sales dataset
data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Laptop", "Mouse"],
    "Quantity": [2, 10, 5, 1, 8],
    "Price": [3000, 50, 120, 3000, 50]
}

# Convert data into a table
df = pd.DataFrame(data)

# Calculate revenue for each product sale
df["Revenue"] = df["Quantity"] * df["Price"]

# Display the sales table
print("SALES DATA")
print(df)

# Calculate total revenue
total_revenue = df["Revenue"].sum()

print("\nTotal Revenue: RM", total_revenue)

# Find the product with the highest total revenue
product_revenue = df.groupby("Product")["Revenue"].sum()

print("\nRevenue by Product:")
print(product_revenue.sort_values(ascending=False))

print("\nMy first data analysis is complete!")
```

```python

import matplotlib.pyplot as plt

# Chart 1: Revenue by product
product_revenue.plot(kind="bar", title="Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (RM)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Chart 2: Quantity sold by product
product_quantity = df.groupby("Product")["Quantity"].sum()

product_quantity.plot(
    kind="bar",
    color="orange",
    title="Quantity Sold by Product"
)
plt.xlabel("Product")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Chart 3: Revenue contribution by product
product_revenue.plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Revenue Share by Product"
)
plt.ylabel("")
plt.show()

print("All three charts created successfully!")
```
